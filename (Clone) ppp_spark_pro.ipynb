```python
from pyspark.sql.functions import col
from pyspark.sql.types import StructType, StructField, IntegerType, StringType, DateType

# Step 1: Load the table (current "header row" is really the first row of real data)
df_raw = spark.read.table("workspace.default.sales")

# Step 2: Rename the columns to something temporary
df_temp = df_raw.toDF("col1", "col2", "col3", "col4", "col5")

# Step 3: Define your schema
schema = StructType([
    StructField("product_id", IntegerType(), True),
    StructField("customer_id", StringType(), True),
    StructField("order_date", DateType(), True),
    StructField("location", StringType(), True),
    StructField("source_order", StringType(), True)
])

# Step 4: Cast and rename columns correctly
df = df_temp.select(
    col("col1").cast(IntegerType()).alias("product_id"),
    col("col2").cast(StringType()).alias("customer_id"),
    col("col3").cast(DateType()).alias("order_date"),
    col("col4").cast(StringType()).alias("location"),
    col("col5").cast(StringType()).alias("source_order")
)

# Step 5: Show final schema and data
df.printSchema()
df.show(2)
```

```python
sales_df = df
display(sales_df)
from pyspark.sql.functions import month, year , quarter

sales_df = sales_df.withColumn("order_year",year(sales_df.order_date))
sales_df = sales_df.withColumn("order_month",month(sales_df.order_date))
sales_df = sales_df.withColumn("order_quarter",quarter(sales_df.order_date))
display(sales_df)
```

```python
from pyspark.sql.types import StructType, StructField, IntegerType, StringType

# Step 1: Read the existing table
df = spark.read.table("workspace.default.menu")

# Step 2: Rename columns (if needed) — assuming table currently has no or incorrect column names
# If columns already exist but you just want to rename them:
df = df.toDF("product_id", "product_name", "price")

# Step 3: Cast to correct data types
df = df.select(
    df["product_id"].cast(IntegerType()),
    df["product_name"].cast(StringType()),
    df["price"].cast(IntegerType())
)

# Optional: View schema
df.printSchema()
menu_df = df
# Optional: Show data
df.show()
```

```python
from pyspark.sql import functions as F

total_amount_spent = (
    sales_df
    .join(menu_df, "product_id")
    .groupBy("customer_id")
    .agg(F.sum("price").alias("total_amount_spent"))
    .orderBy("customer_id")
)

display(total_amount_spent)
```
